## Popcorn Hack 1: report_status

One procedure, called twice with different arguments for its two parameters.

````python
# CODE_RUNNER: Procedures Popcorn #1

def report_status(location, status):
    print(f"Status update from {location}: {status}")

report_status("Poway", "Roads clear")
report_status("Rancho Bernardo", "Crews on scene")
````

**Actual output:**

```text
Status update from Poway: Roads clear
Status update from Rancho Bernardo: Crews on scene
```

## Popcorn Hack 2: decomposition and a helper procedure

`process_report` calls the helper `is_priority`, stores the returned Boolean in `priority`, and uses it in an `if` / `else`.

````python
# CODE_RUNNER: Popcorn Hack 2 - Decomposition & Helper Procedures

# Helper Procedure
def is_priority(incident_type):
    return incident_type == "Medical Emergency"

# Outer Procedure
def process_report(location, incident_type):
    priority = is_priority(incident_type)
    if priority:
        print("PRIORITY: " + incident_type + " reported in " + location)
    else:
        print("STANDARD REPORT: " + incident_type + " reported in " + location)

process_report("Poway", "Medical Emergency")
process_report("Poway", "Power Outage")
````

**Actual output:**

```text
PRIORITY: Medical Emergency reported in Poway
STANDARD REPORT: Power Outage reported in Poway
```

## Homework Hack: emergency response dispatcher

`check_priority` is the helper: it returns `True` for incidents where lives are at risk right now. `dispatch_response` calls it, stores the result, and prints a different response for each case. It is called four times with four incident types and three locations, so the same code handles every report without being copied.

````python
# CODE_RUNNER: Homework Hack - Emergency response dispatcher

# Helper procedure: returns a Boolean
def check_priority(incident_type):
    return incident_type == "Medical Emergency" or incident_type == "Wildfire"

# Outer procedure: calls the helper, then chooses a response
def dispatch_response(location, incident_type):
    priority = check_priority(incident_type)
    if priority:
        print("DISPATCH NOW: sending emergency crews to " + location + " for " + incident_type)
    else:
        print("SCHEDULED: " + incident_type + " in " + location + " added to the work queue")

dispatch_response("Poway", "Medical Emergency")
dispatch_response("Rancho Bernardo", "Power Outage")
dispatch_response("Escondido", "Wildfire")
dispatch_response("Poway", "Downed Tree")
````

**Actual output:**

```text
DISPATCH NOW: sending emergency crews to Poway for Medical Emergency
SCHEDULED: Power Outage in Rancho Bernardo added to the work queue
DISPATCH NOW: sending emergency crews to Escondido for Wildfire
SCHEDULED: Downed Tree in Poway added to the work queue
```

**How this uses procedures:** `def` defines each procedure; `location` and `incident_type` are parameters; the strings in each call (such as `"Poway"`, `"Wildfire"`) are the arguments. Calling one helper from `dispatch_response` means the priority rule lives in one place, so changing which incidents are urgent is a one-line edit.
