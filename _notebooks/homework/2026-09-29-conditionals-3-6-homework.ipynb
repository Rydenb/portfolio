Running the original prints nothing, and that is a bug, not an error: `0 > 100` is False and there was no `else`.

## Popcorn Hack 2

With the original order, score 9 prints "Be a little careful" because `>= 3` is checked first. The `>= 6` and `>= 9` branches can never run. Fix: order thresholds highest first.

```python
# CODE_RUNNER: 3.6 Popcorn 2 - Fix the order of the elif chain

risk_score = 9

# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
```

## Homework Hack: score a message, then advise on it

```python
# CODE_RUNNER: 3.6 Homework - score a message, then advise on it

message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False

risk_score = 0

# Each signal is its own if with no else: when a signal is absent the right
# action is "add nothing", so there is no second answer to give.
if "gift card" in message:      # no else: no gift card means add 0
    risk_score = risk_score + 4
if amount_requested > 100:      # no else: a small amount means add 0
    risk_score = risk_score + 3
if not known_contact:           # no else: a known contact means add 0
    risk_score = risk_score + 2

print("Risk score:", risk_score)

# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
```

### Reachability proof

Possible scores are 0, 2, 3, 4, 5, 6, 7, 9 (sums of 4, 3, 2).

| Branch | Inputs | Score |
|---|---|---|
| `>= 9` Do not reply | gift card, amount 500, unknown sender | 9 |
| `>= 6` Risky | gift card, amount 500, known contact | 7 |
| `>= 3` Be careful | no gift card, amount 500, known contact | 3 |
| `else` Nothing suspicious | no gift card, amount 0, known contact | 0 |

Every branch is reachable, so none is dead code.

**Stretch:** the scoring `if`s are fine without `else` because skipping a signal correctly means "add nothing", whereas in Example A the missing `else` left the user with no message at all.
