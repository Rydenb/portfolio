## Popcorn Hack 1: the silent failure

Running the original with `amount_requested = 0` prints **nothing**. That is a bug, not an error: `0 > 100` is False and there was no `else`, so one of the two outcomes had no answer. The fix adds an `else`.

````python
# CODE_RUNNER: 3.6 Popcorn 1 - Run it first. Then make both outcomes print.
amount_requested = 0

# 1 condition -> 2 outcomes -> 2 answers
if amount_requested > 100:
    print("This message is asking for a large amount of money.")
else:
    print("This message is not asking for a large amount of money.")
````

**Actual output:**

```text
This message is not asking for a large amount of money.
```

## Popcorn Hack 2: the chain in the wrong order

With the original order, score 9 is told **"Be a little careful"**, because `>= 3` is checked first and is already true. The `>= 6` branch and the `>= 9` branch are **unreachable**: any score that could reach them is caught by `>= 3` first. Fix: check the highest threshold first.

````python
# CODE_RUNNER: 3.6 Popcorn 2 - Run it first. Then fix the order.
risk_score = 9

# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
````

**Actual output:**

```text
Do not reply. This is almost certainly a scam.
```

## Homework: build the trainer's scoring step

Each signal is a separate `if` (gift cards +4, amount over 100 +3, unknown sender +2). The gift-card check uses `message.lower()` so "Gift Cards" and "gift cards" both count. The report is an `if` / `elif` / `else` chain, highest threshold first, with the count comment above it.

````python
# CODE_RUNNER: 3.6 Homework - score a message, then advise on it
message = 'URGENT: send gift cards now to unlock your account'
amount_requested = 500
known_contact = False

risk_score = 0

# Each signal is its own if with no else: when a signal is absent the right
# action is "add nothing", so there is no second answer to give.
if "gift card" in message.lower():   # no else: no gift card means add 0
    risk_score = risk_score + 4
if amount_requested > 100:           # no else: a small amount means add 0
    risk_score = risk_score + 3
if not known_contact:                # no else: a known contact means add 0
    risk_score = risk_score + 2

print("Risk score:", risk_score)

# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
````

**Actual output:**

```text
Risk score: 9
Do not reply. This is almost certainly a scam.
```
