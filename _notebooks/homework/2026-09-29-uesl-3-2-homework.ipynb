## Homework

### Grow your UESL game

Refactor 3.1 into `level_stars = [12, 18, 10]`. Each element is one level's collected stars, in level order (index 0 is level one, index 1 is level two, index 2 is level three). Then add 5 to the second level, append 7 for a new level, and use the traversal (Example C) to print the collection, level count and total.

**Prediction:** `[12, 23, 10, 7]`, `4`, `52` (12 + 23 + 10 + 7).

````python
# CODE_RUNNER: UESL 3.2 Homework - Grow your UESL game
level_stars = [12, 18, 10]            # each element = one level's stars, in level order
if len(level_stars) >= 2:             # only update the second level if it exists
    level_stars[1] = level_stars[1] + 5
level_stars.append(7)                 # a new fourth level

total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print(level_stars)
print(len(level_stars))
print(total_stars)
````

**Actual output:**

```text
[12, 23, 10, 7]
4
52
```

This matched my prediction.

**College Board pseudocode version** (AP indices start at 1, so the second level is index 2):

```text
level_stars ← [12, 18, 10]
level_stars[2] ← level_stars[2] + 5
APPEND(level_stars, 7)
total_stars ← 0
FOR EACH stars IN level_stars
{
    total_stars ← total_stars + stars
}
DISPLAY(level_stars)
DISPLAY(LENGTH(level_stars))
DISPLAY(total_stars)
```

**Edits a list avoids:** with 100 separate variables, every new level needs a new variable name *and* an edit to the sum line (`level_1 + level_2 + ... + level_101`); forgetting one gives a silently wrong total. With a list, `append` adds a level and the same three-line traversal totals any number of levels, including none.

**Level labels:** players should see names like Forest, Cavern and Summit on screen instead of "index 1". A maker can keep a parallel list of labels in the same order, so position `i` in both lists describes the same level.

## Design Thinking

- **Player/maker need:** A maker wants to add or reorder levels without losing track of each level's stars, and a player should never need to understand indices to see their progress.
- **Goal:** One understandable collection in level order that can be read, updated and extended, and still totals correctly with zero levels.
- **Representation I considered:** Separate variables per level (the 3.1 approach) and a dictionary keyed by level name like `{"Forest": 12}`. I chose a plain list because order matters and the traversal is simple; every element stays numeric so addition makes sense.
- **Prototype:** `level_stars = [12, 18, 10]` with an update on index 1, an append, and the traversal total.
- **Test and revision:** My first draft of the empty test ran `level_stars[1] = level_stars[1] + 5` directly, which raises `IndexError` on an empty list. I added the `len(level_stars) >= 2` guard and retested to get `0`, `0`.
