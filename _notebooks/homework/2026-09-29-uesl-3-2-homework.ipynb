## Popcorn

**Prediction:** `level_stars` starts as `[6, 9, 12]`. Index 1 is the second level, so `9 + 3 = 12` replaces it, giving `[6, 12, 12]`. Then `append(5)` adds a fourth element: `[6, 12, 12, 5]`, and the length is `4`.

**AP index:** Python index 1 is AP index **2**, since AP lists start at 1.

````python
# Popcorn Hack: extend the level sequence
level_stars = [6, 9, 12]
level_stars[1] = level_stars[1] + 3
level_stars.append(5)
print(level_stars)
print(len(level_stars))
````

**Actual output:** `[6, 12, 12, 5]`, then `4`. This matched my prediction.

**Changed version:** appending `8` instead gives `[6, 12, 12, 8]`, still 4 levels. Replacing an element overwrites a slot, so the length is unchanged (`9` became `12`). Appending adds a new slot at the end, so the length grows by 1.

**Adapted Example C (my own values):** total a list of four levels, then an empty project.

````python
# Adapted Example C: traversal total for my own list
level_stars = [7, 9, 4, 11]
total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars
print(len(level_stars))
print(total_stars)
````

**Prediction:** `4`, then `7 + 9 + 4 + 11 = 31`. **Actual output:** `4`, `31`. Matched. The loop visits every element, so it works for any length with no edits.

## Homework

Refactor the 3.1 prototype into `level_stars = [12, 18, 10]`. Each element is one level's collected stars, in level order (index 0 is level one, index 1 is level two, index 2 is level three). Then add 5 to the second level, append 7 for a new level, and use the traversal to print the collection, level count and total.

**Prediction:** `[12, 23, 10, 7]`, `4`, `52` (12 + 23 + 10 + 7).

````python
# Homework Hack: grow the UESL game
level_stars = [12, 18, 10]       # each element = one level's stars, in level order
level_stars[1] = level_stars[1] + 5   # bonus of 5 on the second level
level_stars.append(7)                 # a new fourth level

total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print(level_stars)
print(len(level_stars))
print(total_stars)
````

**Actual output:** `[12, 23, 10, 7]`, `4`, `52`. This matched my prediction.

**College Board pseudocode version** (AP indices start at 1, so the second level is index 2):

```text
level_stars ← [12, 18, 10]
level_stars[2] ← level_stars[2] + 5
APPEND(level_stars, 7)
total_stars ← 0
FOR EACH stars IN level_stars
{
    total_stars ← total_stars + stars
}
DISPLAY(level_stars)
DISPLAY(LENGTH(level_stars))
DISPLAY(total_stars)
```

**Connection to 3.1:** changing a list still does not recalculate a total that was stored earlier. The list organizes the state; the assignment and the traversal are what update it. That is why `total_stars` is computed *after* the update and append.

### Test 2 · Empty project

**Prediction:** `[]`, `0`, `0`. The loop body never runs, so `total_stars` stays at its starting value of 0. The second-level update is skipped because an empty list has no index 1 (it would raise `IndexError`), so the code guards it with `if len(level_stars) >= 2`.

````python
# Test 2: empty project
level_stars = []
if len(level_stars) >= 2:             # skip the update when there is no second level
    level_stars[1] = level_stars[1] + 5
total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars
print(level_stars)
print(len(level_stars))
print(total_stars)
````

**Actual output:** `[]`, `0`, `0`. Pass.

| Test | Starting list | Final list | Levels | Total | Result |

| 1 | `[12, 18, 10]` | `[12, 23, 10, 7]` | 4 | 52 | Pass |
| 2 | `[]` | `[]` | 0 | 0 | Pass |

**Complexity explanation:** with separate variables, 100 levels would need 100 variable names, and every new level would need a new variable *and* an edit to the sum line (`level_1 + level_2 + ... + level_101`). Forgetting one gives a silently wrong total. With a list, one name holds every level, `append` adds a level, and the same 3-line traversal totals 3 levels, 4 levels, or none. The repeated edits are eliminated, which is how the abstraction reduces complexity.

## Design Thinking

- **Player/maker need:** A maker wants to add or reorder levels without losing track of each level's stars, and a player should never need to understand indices to see their progress.
- **Goal:** One understandable collection in level order that can be read, updated and extended, and still totals correctly with zero levels.
- **Representation I considered:** Separate variables per level (the 3.1 approach) and a dictionary keyed by level name like `{"Forest": 12}`. I chose a plain list because order matters and the traversal is simple. Every element stays numeric so addition makes sense.
- **Prototype:** `level_stars = [12, 18, 10]` with an update on index 1, an append, and the traversal total.
- **Test and revision:** My first draft of the empty test ran `level_stars[1] = level_stars[1] + 5` directly, which raises `IndexError` on an empty list. I added the `len(level_stars) >= 2` check and retested to get `[]`, `0`, `0`.
- **Level labels:** Players should see names like Forest, Cavern and Summit on screen instead of "level 0" or "index 1". A maker could keep a parallel list of labels in the same order so position `i` in both lists describes the same level.
