The lesson writes single-underscore classes (`ocs_lead`, `ocs_card`, `ocs_section-title`), but those do not exist in the SCSS. The real classes are `ocs__lead`, `ocs__card` and `ocs__section-title` (double underscore), so I used those.

## Popcorn Hack

Removed the `<style>` block and the custom classes. The fake heading `div` became `<h1>`, the `span` subtitle became `<h2>`, and the text became `<p class="ocs__lead">`.

## Homework Hack

Removed the `<style>` block and every custom class. The title is an `h2.ocs__section-title`, the subtitle an `h3`, the text is in `<p>` tags, the bold label is `<strong>`, the three hand-numbered lines are one real `<ol>` with three `<li>` (so the manual "1." "2." "3." are gone), and the closing line is an `h3`. Everything sits in an `ocs__card`.

{% capture typography_homework_challenge %}
Homework: rebuild How to Pat a Cat with semantic headings, paragraphs, strong and a real ordered list.
{% endcapture %}

{% capture typography_homework_code %}
outputElement.innerHTML = [
  '<div class="ocs__card">',
  '  <h2 class="ocs__section-title">How to Pat a Cat</h2>',
  '  <h3>Have you ever wanted to...</h3>',
  '  <p>pat a cat? Well good for you we will give step by step instructions on how to pat a cat!</p>',
  '  <p><strong>How to pat a cat:</strong></p>',
  '  <ol>',
  '    <li>put out your hand</li>',
  '    <li>touch the cat</li>',
  '    <li>move your hand back and forth</li>',
  '  </ol>',
  '  <h3>Congrats! You pat a cat :D</h3>',
  '</div>'
].join('');
{% endcapture %}
