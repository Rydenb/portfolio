## Popcorn Hack 1

Basic arithmetic operators with `value_a`, `value_b`, and `value_c`.

````python
# CODE_RUNNER: Popcorn Hack - Basic Arithmetic Operators

value_a = 100
value_b = 10
value_c = 25

# total of A, B, and C
addition = value_a + value_b + value_c

# difference between A and B
subtraction = value_a - value_b

# product of B and C
multiplication = value_b * value_c

# A divided by B (/ always gives a float)
division = value_a / value_b

print("Total:", addition)
print("Difference:", subtraction)
print("Product:", multiplication)
print("Division:", division)
````

**Actual output:**

```text

Product: 250
Division: 10.0
```

## Popcorn Hack 2

### Part A: Python version of the booking pseudocode

Grouping operations for the Poway Recovery Center booking system.

````python
# CODE_RUNNER: Python Version of Popcorn Hack 2

total_monthly_capacity = 100
spots_requested_per_group = 4
sessions_booked = 3

# 1. Multiplication: total spots taken by the booking group
total_spots_taken = spots_requested_per_group * sessions_booked

# 2. Subtraction: remaining capacity for the center
remaining_capacity = total_monthly_capacity - total_spots_taken

# 3. Division: distribute remaining spots evenly across 2 active rooms
users_per_room = remaining_capacity / 2

print("Total Spots Taken:", total_spots_taken)
print("Remaining Capacity:", remaining_capacity)
print("Users per Room:", users_per_room)
````

**Actual output:**

```text
Total Spots Taken: 12
Remaining Capacity: 88
Users per Room: 44.0
```

## AP CSP Practice Questions

**1. Final values of `a`, `b`, and `c`**: **B) a = 8, b = 6, c = 14**

Starting from `a ← 4`, `b ← 6`, `c ← 2`:
- `a ← b + c` makes `a` = 6 + 2 = 8
- `b ← a - c` makes `b` = 8 - 2 = 6 (it uses the *new* `a`)
- `c ← a + b` makes `c` = 8 + 6 = 14

**2. `result ← 10 + 17 MOD 5 * 2`**: **A) 14**

`MOD` and `*` have the same precedence and go left to right before `+`:
- `17 MOD 5` = 2
- `2 * 2` = 4
- `10 + 4` = 14

**3. Which expression evaluates to 3.5?**: **B) 7 / 2**

In AP CSP pseudocode, `/` gives the exact decimal result, so `7 / 2` = 3.5. The others: A) `7 MOD 2 + 2` = 3, C) `14 MOD 4 / 2` = 1, D) `3 + 1 / 2 * 0` = 3.

Checking all three in Python (`%` is Python's `MOD`):

````python
# CODE_RUNNER: Check the practice question answers

# Question 1
a = 4
b = 6
c = 2
a = b + c
b = a - c
c = a + b
print("Q1: a =", a, " b =", b, " c =", c)

# Question 2
result = 10 + 17 % 5 * 2
print("Q2: result =", result)

# Question 3
print("Q3: A =", 7 % 2 + 2, " B =", 7 / 2, " C =", 14 % 4 / 2, " D =", 3 + 1 / 2 * 0)
````

**Actual output:**

```text
Q1: a = 8  b = 6  c = 14
Q2: result = 14
Q3: A = 3  B = 3.5  C = 1.0  D = 3.0
```

## Homework: Recovery Center Cost Report

The enrollment fee is a one-time cost, so it is added to the monthly cost once. The average daily cost uses a 30-day month.

````python
# CODE_RUNNER: HOMEWORK

# Recovery Center Cost Report

cost_per_session = 35
sessions_per_week = 3
weeks_per_month = 4
enrollment_fee = 75

# 1. Calculate weekly cost
weekly_cost = cost_per_session * sessions_per_week

# 2. Calculate monthly cost
monthly_cost = weekly_cost * weeks_per_month

# 3. Calculate total cost with fee
total_cost_with_fee = monthly_cost + enrollment_fee

# 4. Calculate average daily cost
average_daily_cost = monthly_cost / 30

# Print all results with labels
print("Weekly Cost:", weekly_cost)
print("Monthly Cost:", monthly_cost)
print("Total Cost with Fee:", total_cost_with_fee)
print("Average Daily Cost:", average_daily_cost)
````

**Actual output:**

```text
Weekly Cost: 105
Monthly Cost: 420
Total Cost with Fee: 495
Average Daily Cost: 14.0
```
