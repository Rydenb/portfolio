## Popcorn Hack

Removed the inline style and the `box` class. The search field is `ocs__input large gradient` and the first-name field is a compact `ocs__input small`. Each has a visible label.

{% capture inputs_popcorn_challenge %}
Popcorn: one large gradient input and one compact plain input, using OCS input grammar.
{% endcapture %}

{% capture inputs_popcorn_code %}
outputElement.innerHTML = [
  '<label for="pvo-search">Search PVO resources</label>',
  '<input id="pvo-search" type="text" class="ocs__input large gradient" placeholder="Search PVO resources...">',
  '<label for="pvo-first-name">Volunteer First Name</label>',
  '<input id="pvo-first-name" type="text" class="ocs__input small" placeholder="Volunteer First Name">'
].join('');
{% endcapture %}
