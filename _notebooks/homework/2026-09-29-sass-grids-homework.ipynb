## MCQ

1. Four boxes per row in a standard grid: **A** `cols-4`.
2. `grid-template-columns: 2fr 1fr` makes **B** 2 boxes, the first twice as wide.
3. A cell spanning the whole row: **C** `ocs__grid-cell--header`.
4. The line that breaks the rule is **D** `<div class="my-row" style="display:flex">`: a made-up class plus an inline style.
