### Library recommendation: 1,000 books

````python
# CODE_RUNNER: Library 3.17 Homework - Checkpoint B with 1000 books
import math

n = 1000
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1
print("Books:", n)
print("Linear search worst case:", linear_checks)
print("Binary search worst case:", binary_checks)
````

**Actual output:**

```text
Books: 1000
Linear search worst case: 1000
Binary search worst case: 10
```

**Recommendation:** 1,000 worst-case checks for linear search versus 10 for binary search, so for **repeated searches of an already sorted catalog, use binary search**.

If the catalog is **unsorted**, binary search cannot be used until it is sorted, and sorting costs extra work up front. Keeping an **extra sorted copy** avoids re-sorting but costs extra memory. That time/memory tradeoff is worth it when the librarian searches many times; for a single search, one linear scan is cheaper than sorting first.
