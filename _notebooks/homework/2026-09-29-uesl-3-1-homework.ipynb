## Popcorn

### Checkpoint challenge

**Prediction:** `stars` starts at 8, `checkpoint_stars` copies 8, then `stars` becomes 8 + 4 = 12. The checkpoint was not reassigned, so it stays 8. Prediction: `12`, then `8`.

````python
# CODE_RUNNER: UESL 3.1 Popcorn - Keep the checkpoint
stars = 8
checkpoint_stars = stars
stars = stars + 4
print(stars)
print(checkpoint_stars)
````

**Actual output:**

```text
12
8
```

This matched my prediction.

**Why they differ:** `checkpoint_stars = stars` copies the *value* that `stars` held at that moment (8). It does not link the two names together. When `stars = stars + 4` runs, Python evaluates the right side (8 + 4 = 12) and replaces only the value of `stars`. The checkpoint keeps the snapshot of 8, which is exactly what a checkpoint should do.

## Homework

### Build your UESL star counter

A game name, a reduced-motion Boolean setting, three level-star variables starting at 12, 18 and 10, and a bonus of 5 added to level two.

**Prediction:** `Comet Collector`, `True`, `40` (original), `40` (stored total after the bonus, not recalculated yet), `45` (recalculated).

````python
# CODE_RUNNER: UESL 3.1 Homework - Build your UESL star counter
game_title = "Comet Collector"   # str: the game's name is text
reduced_motion = True            # bool: the player turned reduced motion on
level_one_stars = 12             # int: stars are whole items
level_two_stars = 18
level_three_stars = 10
bonus = 5                        # int: bonus stars earned on level two

total_stars = level_one_stars + level_two_stars + level_three_stars
print(game_title)
print(reduced_motion)
print(total_stars)               # original total

level_two_stars = level_two_stars + bonus
print(total_stars)               # stored total: not updated yet

total_stars = level_one_stars + level_two_stars + level_three_stars
print(total_stars)               # recalculated total
````

**Actual output:**

```text
Comet Collector
True
40
40
45
```

This matched my prediction: original/stored total 40, updated total 45.

**College Board pseudocode version** (same values, same order):

```text
game_title ← "Comet Collector"
reduced_motion ← true
level_one_stars ← 12
level_two_stars ← 18
level_three_stars ← 10
bonus ← 5
total_stars ← level_one_stars + level_two_stars + level_three_stars
DISPLAY(game_title)
DISPLAY(reduced_motion)
DISPLAY(total_stars)
level_two_stars ← level_two_stars + bonus
DISPLAY(total_stars)
total_stars ← level_one_stars + level_two_stars + level_three_stars
DISPLAY(total_stars)
```

## Tests

Each test starts from fresh values, so nothing carries over from an earlier runner.

### Test 1: bonus = 5

### Test 2: bonus = 0

**Prediction:** `40`, `40`, `40`.

````python
# CODE_RUNNER: UESL 3.1 Test 2 - bonus = 0
level_one_stars = 12
level_two_stars = 18
level_three_stars = 10
bonus = 0

total_stars = level_one_stars + level_two_stars + level_three_stars
print(total_stars)
level_two_stars = level_two_stars + bonus
print(total_stars)
total_stars = level_one_stars + level_two_stars + level_three_stars
print(total_stars)
````

**Actual output:**

```text
40
40
40
```

Pass. With a bonus of 0, `level_two_stars` is reassigned to 18 + 0 = 18, so recalculating gives the same total. This control case shows the jump to 45 in Test 1 came from the bonus, not from the recalculation itself.

| Test | Bonus | Original total | Stored total after bonus | Recalculated total | Result |
|------|-------|----------------|--------------------------|--------------------|--------|
| 1 | 5 | 40 | 40 | 45 | Pass |
| 2 | 0 | 40 | 40 | 40 | Pass |

**Assignment vs. comparison:** `total_stars = ...` in Python (or `total_stars ← ...` in AP pseudocode) evaluates the right side, then stores the result in the name on the left. `total_stars == 40` in Python (or `total_stars = 40` in AP pseudocode) only asks whether the value equals 40 and produces `True` or `False`; it changes nothing.

## Design Thinking

- **Player need:** Some players find fast animation, screen shake or flashing effects uncomfortable or distracting, and every player needs to trust that the star count they see is correct after earning a bonus.
- **Goal:** Store the player's settings and progress clearly, so a maker can trace every change to the star count and a stored total never gets confused with an updated one.
- **Setting/representation I considered:** I thought about storing reduced motion as a string like `"on"`/`"off"`, but a Boolean fits better because the setting has exactly two states and can be checked directly later (`if reduced_motion:`). Each level's stars is an `int` because stars are whole items, and the game title is a `str`.
- **Prototype:** The Comet Collector star counter above: a title, a reduced-motion setting, three level variables, a bonus, and a total printed before and after recalculation.
- **Test / revision:** My first draft hard-coded `+ 5` on level two. To run the bonus-0 test I would have had to edit the math line itself, so I revised it to use a `bonus` variable. Each test now changes one clearly labelled value, and Test 2 (bonus 0) confirmed the recalculation logic on its own.
- **Why reduced motion should be a player choice:** Players' needs differ. Motion one player enjoys can make another feel sick or lose focus, and the player is the one who knows what works for them. A setting the player controls lets each person play comfortably without taking the effect away from people who like it.
