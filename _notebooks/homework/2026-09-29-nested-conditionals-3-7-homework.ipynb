## Popcorn Hack 1

**Prediction:** `Finish homework`. Outer branch: `is_weekday` True. Inner branch: `homework_finished` False. Other outcomes: finished -> `Free time`, weekend -> `Weekend plan`.

````python
# CODE_RUNNER: Popcorn Hack 1 - Test the after-school homework plan

def plan(is_weekday, homework_finished):
    if is_weekday:
        if homework_finished:
            return "Free time"
        else:
            return "Finish homework"
    else:
        return "Weekend plan"

print("weekday, not finished:", plan(True, False))
print("weekday, finished:", plan(True, True))
print("weekend:", plan(False, False))
````

## Popcorn Hack 2

The original was not nested because the second `if` was at the same indentation as the first, so permission was checked even when not logged in. Fixed by indenting it inside `if logged_in` and adding an outer `else`.

````python
# CODE_RUNNER: Popcorn Hack 2 - Repair the student account check

def check(logged_in, has_permission):
    if logged_in:
        print("Account found")
        if has_permission:
            print("Page opened")
        else:
            print("Permission needed")
    else:
        print("Log in first")

check(True, True)
print("--")
check(True, False)
print("--")
check(False, False)
````

## Popcorn Hack 3

The outer condition, `lunch_packed`, is checked first; `practice_after_school` is only checked inside each branch.

JavaScript:
```js
if (lunchPacked) {
  if (practice) { /* packed + snack */ } else { /* packed */ }
} else {
  if (practice) { /* buy + snack */ } else { /* buy */ }
}
```
College Board pseudocode:
```
IF (lunchPacked) { IF (practice) { DISPLAY("packed + snack") } ELSE { DISPLAY("packed") } }
ELSE { IF (practice) { DISPLAY("buy + snack") } ELSE { DISPLAY("buy") } }
```

````python
# CODE_RUNNER: Popcorn Hack 3 - Test the lunch and sports-practice plan

def lunch_plan(lunch_packed, practice_after_school):
    if lunch_packed:
        if practice_after_school:
            return "Eat your packed lunch and bring an extra snack for practice"
        else:
            return "Eat your packed lunch"
    else:
        if practice_after_school:
            return "Buy lunch and buy an extra snack for practice"
        else:
            return "Buy lunch"

for packed in [True, False]:
    for practice in [True, False]:
        print("packed:", packed, "| practice:", practice, "->", lunch_plan(packed, practice))
````

## Homework Hack: Daily Routine Dashboard

````python
# CODE_RUNNER: Homework Hack - Build a high school daily routine dashboard

student = {
    "logged_in": True,
    "homework_finished": False,
    "lunch_packed": True,
    "practice_today": True
}

def dashboard(student):
    if student["logged_in"]:
        # homework decision only happens once the student is logged in
        if student["homework_finished"]:
            if student["practice_today"]:
                print("Homework done: go to practice, then dinner")
            else:
                print("Homework done: enjoy free time")
        else:
            if student["practice_today"]:
                print("Homework first, then practice - bring a snack")
            else:
                print("Finish homework before dinner")
        # meal decision, also only for logged-in students
        if student["lunch_packed"]:
            print("Meal: eat your packed lunch")
        else:
            print("Meal: buy lunch at school")
    else:
        print("Log in to see your dashboard")

# test every meaningful outcome
tests = [
    dict(student, logged_in=False),
    dict(student, homework_finished=True, practice_today=True),
    dict(student, homework_finished=True, practice_today=False, lunch_packed=False),
    dict(student, homework_finished=False, practice_today=True),
    dict(student, homework_finished=False, practice_today=False, lunch_packed=False),
]
for t in tests:
    print(t)
    dashboard(t)
    print()
````

### Why the inner checks depend on the outer decisions

Homework, practice and meals only matter once the student is logged in, so those checks sit inside the `logged_in` branch and are skipped entirely when it is False. Likewise, whether practice is today only changes the advice after we know if homework is finished. Indentation decides which `if` each `else` belongs to, so moving one would change the outcome.
