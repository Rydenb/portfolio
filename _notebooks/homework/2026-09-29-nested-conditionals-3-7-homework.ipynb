## Popcorn Hack 1: after-school homework plan

**Trace:** with `is_weekday = True` and `homework_finished = False`, the **outer** branch taken is `if is_weekday` and the **inner** branch is the `else`, so it prints **`Finish homework`**.

**Other outcomes:** `homework_finished = True` reaches `Free time`; `is_weekday = False` reaches `Weekend plan` (the inner check is skipped).

````python
# CODE_RUNNER: Popcorn Hack 1 - Test the after-school homework plan

def plan(is_weekday, homework_finished):
    if is_weekday:
        if homework_finished:
            print("Free time")
        else:
            print("Finish homework")
    else:
        print("Weekend plan")

plan(True, False)    # the starter values
plan(True, True)     # homework finished
plan(False, False)   # weekend
````

**Actual output:**

```text
Finish homework
Free time
Weekend plan
```

## Popcorn Hack 2: repair the account check

The starter was **not nested**: `if has_permission` sat at the same indentation as `if logged_in`, so permission was checked even for someone who was not logged in. Fix: move the permission check inside the logged-in branch and add an outer `else` that prints "Log in first".

````python
# CODE_RUNNER: Popcorn Hack 2 - Repair the student account check

def check(logged_in, has_permission):
    if logged_in:
        if has_permission:
            print("Page opened")
        else:
            print("Permission needed")
    else:
        print("Log in first")

check(True, True)     # Page opened
check(True, False)    # the starter values: Permission needed
check(False, False)   # Log in first
````

**Actual output:**

```text
Page opened
Permission needed
Log in first
```

## Popcorn Hack 3: lunch and practice

`lunch_packed` is checked **first** (the outer condition); `practice_after_school` is checked only inside each branch, which gives 4 recommendations.

````python
# CODE_RUNNER: Popcorn Hack 3 - Test the lunch and sports-practice plan

def lunch_plan(lunch_packed, practice_after_school):
    if lunch_packed:
        if practice_after_school:
            print("Eat your packed lunch and bring an extra snack for practice")
        else:
            print("Eat your packed lunch")
    else:
        if practice_after_school:
            print("Buy lunch and buy an extra snack for practice")
        else:
            print("Buy lunch")

lunch_plan(True, True)
lunch_plan(True, False)
lunch_plan(False, True)
lunch_plan(False, False)
````

**Actual output:**

```text
Eat your packed lunch and bring an extra snack for practice
Eat your packed lunch
Buy lunch and buy an extra snack for practice
Buy lunch
```

## Homework Hack: High School Daily Routine Dashboard

Four student-life values (login, homework, meal, practice). The account check is the outer `if` / `else`, before any schedule checks. Inside it there are two nested decisions (homework → practice, and meal), giving five different homework/practice/login recommendations plus two meal recommendations.

````python
# CODE_RUNNER: Homework Hack - Build a high school daily routine dashboard
student = {"logged_in": True, "homework_finished": False, "lunch_packed": True, "practice_today": True}

def dashboard(student):
    # outer decision: nothing else matters until the account check passes
    if student["logged_in"]:
        # nested decision 1: homework, then practice inside each homework branch
        if student["homework_finished"]:
            if student["practice_today"]:
                print("Homework done: go to practice, then dinner")
            else:
                print("Homework done: enjoy free time")
        else:
            if student["practice_today"]:
                print("Homework first, then practice - bring a snack")
            else:
                print("Finish homework before dinner")
        # nested decision 2: meal, also only for logged-in students
        if student["lunch_packed"]:
            print("Meal: eat your packed lunch")
        else:
            print("Meal: buy lunch at school")
    else:
        print("Log in to see your dashboard")

dashboard(student)
````

**Actual output:**

```text
Homework first, then practice - bring a snack
Meal: eat your packed lunch
```

### Why the inner checks depend on the outer decisions

Homework, practice and meals only matter once the student is logged in, so those checks sit inside the `logged_in` branch and are skipped entirely when it is False (the not-logged-in test prints one line). Whether practice is today only changes the advice after we know whether homework is finished, so that check is nested inside each homework branch. Indentation decides which `if` each `else` belongs to, so moving one line would change the outcome.
