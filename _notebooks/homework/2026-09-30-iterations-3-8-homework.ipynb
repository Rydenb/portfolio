## Popcorn Hack: wildfire dispatch scanner

Every blank filled in. My thresholds: **safe below 3**, **critical at 8 or more**.

**Prediction:** 1 and 2 are skipped; 4, 6, 3 need action; 9 is critical, so the scan stops before 5. Safe 2, action 4, total 4 + 6 + 3 + 9 = 22, critical found True.

````python
# CODE_RUNNER: Popcorn Hack - wildfire dispatch scanner
alerts = [1, 4, 2, 6, 3, 9, 5]
safe_cutoff = 3
critical_cutoff = 8

action_count = 0
safe_count = 0
total_action_score = 0
critical_found = False

for alert in alerts:
    if alert < safe_cutoff:
        safe_count += 1
        print("Alert " + str(alert) + ": safe, skipping")
        continue
    action_count += 1
    total_action_score += alert
    print("Alert " + str(alert) + ": action needed")
    if alert >= critical_cutoff:
        critical_found = True
        print("Critical alert found - stopping scan")
        break

print("Safe alerts skipped: " + str(safe_count))
print("Action alerts: " + str(action_count))
print("Total action score: " + str(total_action_score))
print("Critical found: " + str(critical_found))
````

**Actual output:**

```text

Alert 4: action needed
Alert 2: safe, skipping
Alert 6: action needed
Alert 3: action needed
Alert 9: action needed
Critical alert found - stopping scan
Safe alerts skipped: 2
Action alerts: 4
Total action score: 22
Critical found: True
```

Matched my prediction. `continue` skipped the rest of the pass for the safe alerts, and `break` ended the loop at 9, so alert 5 was never checked.

## Homework Hack: server room load checker

My own theme: CPU loads for a school server room. 10 values, a safe cutoff (40%) and a critical cutoff (90%), a `for` loop with `continue` and `break`, and 5 tracked summary values. The overloaded server (93%) comes before the end of the list, so `break` actually triggers and the last three servers are never checked.

````python
# CODE_RUNNER: Homework Hack - themed loop-based checker
# Theme: a school server room. Each value is one server's CPU load (%).
cpu_loads = [22, 35, 61, 48, 15, 77, 93, 54, 30, 40]
safe_cutoff = 40         # below 40% the server is fine, skip it
critical_cutoff = 90     # 90% or more means the room is overloaded, stop checking

servers_checked = 0
servers_needing_action = 0
servers_skipped = 0
total_action_load = 0
overload_found = False

for load in cpu_loads:
    servers_checked += 1
    if load < safe_cutoff:
        servers_skipped += 1
        print("Server at " + str(load) + "%: safe, no action")
        continue
    servers_needing_action += 1
    total_action_load += load
    print("Server at " + str(load) + "%: action needed")
    if load >= critical_cutoff:
        overload_found = True
        print("Critical load found. Stop checking.")
        break

print("Final Report")
print("Checked: " + str(servers_checked))
print("Skipped: " + str(servers_skipped))
print("Actions: " + str(servers_needing_action))
print("Total action load: " + str(total_action_load))
print("Overload found: " + str(overload_found))
````

**Actual output:**

```text

Server at 35%: safe, no action
Server at 61%: action needed
Server at 48%: action needed
Server at 15%: safe, no action
Server at 77%: action needed
Server at 93%: action needed
Critical load found. Stop checking.
Final Report
Checked: 7
Skipped: 3
Actions: 4
Total action load: 279
Overload found: True
```
