## Popcorn Hack 1

Only indexing and slicing, no string methods.

````python
# CODE_RUNNER: Popcorn Hack 1 - Extract the hidden message using indexing and slicing

transmission = "xxSIGNALxxxCONFIRMEDxx77xxDELTA"

# SIGNAL starts at index 2 and is 6 characters long
word_1 = transmission[2:8]
print("Word 1:", word_1)

# CONFIRMED comes right after "SIGNALxxx"
word_2 = transmission[11:20]
print("Word 2:", word_2)

# negative slice: the last 5 characters
word_3 = transmission[-5:]
print("Word 3:", word_3)

print("Total length:", len(transmission))

# step of 2 takes every other character
print("Every other character:", transmission[::2])
````

**Actual output:**

```text
Word 1: SIGNAL
Word 2: CONFIRMED
Word 3: DELTA
Total length: 31
Every other character: xSGAxxOFREx7xDLA
```

## Popcorn Hack 2

Only string methods.

````python
# CODE_RUNNER: Popcorn Hack 2 - Clean and search a noisy transmission using string methods

noisy = "   ...AGENT>>>Falcon---reports: the PACKAGE has ARRIVED!!!   "

trimmed = noisy.strip()
print("Trimmed:", repr(trimmed))

cleaned = trimmed.replace(">", "").replace("-", "").replace("!", "")
print("Cleaned:", cleaned)

lowered = cleaned.lower()
print("Lowered:", lowered)

has_arrived = "arrived" in lowered
print("Contains 'arrived':", has_arrived)

words = cleaned.split()
print("Word count:", len(words))
````

**Actual output:**

```text
Trimmed: '...AGENT>>>Falcon---reports: the PACKAGE has ARRIVED!!!'
Cleaned: ...AGENTFalconreports: the PACKAGE has ARRIVED
Lowered: ...agentfalconreports: the package has arrived
Contains 'arrived': True
Word count: 5
```

## Popcorn Hack 3

````python
# CODE_RUNNER: Popcorn Hack 3 - Build a decoded report using concatenation and formatting

agent_name = "Falcon"
target_word = "MIDNIGHT"
confidence = 87          # a percentage, stored as an int
fragments = ["ALPHA", "SEVEN", "DELTA"]

# concatenation with +
headline = "Agent " + agent_name + " intercepted: " + target_word
print(headline)

# str() turns the int into text so + can join it
confidence_line = "Confidence: " + str(confidence) + "%"
print(confidence_line)

# the same line with an f-string
confidence_line_fstring = f"Confidence: {confidence}%"
print(confidence_line_fstring)

joined_fragments = "-".join(fragments)
print("Joined fragments:", joined_fragments)

final_report = f"REPORT: {headline} | {confidence_line} | Fragments: {joined_fragments}"
print(final_report)
````

**Actual output:**

```text
Agent Falcon intercepted: MIDNIGHT
Confidence: 87%
Confidence: 87%
Joined fragments: ALPHA-SEVEN-DELTA
REPORT: Agent Falcon intercepted: MIDNIGHT | Confidence: 87% | Fragments: ALPHA-SEVEN-DELTA
```

## Homework Hack: Crack the Final Transmission

````python
# CODE_RUNNER: Homework - Crack the Final Transmission using slicing, methods, and formatting

raw_transmission = "  2026-09-15,GHOST,THE-EAGLE-LANDS-AT-DAWN,87  "

# TODO 1: strip the spaces, then split on commas into exactly 4 fields
segments = raw_transmission.strip().split(",")
print("Segments:", segments)

# TODO 2: pull each field out by index
date = segments[0]
agent = segments[1]
message_code = segments[2]
confidence_str = segments[3]
print("Date:", date, "| Agent:", agent, "| Message code:", message_code, "| Confidence:", confidence_str)

# TODO 3: slicing (not a method) takes the first 4 characters, the year
year_check = date[0:4] == "2026"
print("Year check:", year_check)

# TODO 4: dashes become spaces
message = message_code.replace("-", " ")
print("Message:", message)

# TODO 5: case-insensitive search
found_dawn = "dawn" in message.lower()
print("Found 'dawn':", found_dawn)

# TODO 6: Markdown table row
markdown_row = f"| {date} | {agent} | {message} | {confidence_str}% |"
print("Markdown row:", markdown_row)

# TODO 7: JSON-style line built with an f-string (no json import).
# The outer curly braces are added as plain strings, so the f-string only holds the fields.
json_line = "{" + f'"date": "{date}", "agent": "{agent}", "message": "{message}", "confidence": {confidence_str}, "contains_dawn": {str(found_dawn).lower()}' + "}"
print("JSON line:", json_line)
````

**Actual output:**

```text
Segments: ['2026-09-15', 'GHOST', 'THE-EAGLE-LANDS-AT-DAWN', '87']
Date: 2026-09-15 | Agent: GHOST | Message code: THE-EAGLE-LANDS-AT-DAWN | Confidence: 87
Year check: True
Message: THE EAGLE LANDS AT DAWN
Found 'dawn': True
Markdown row: | 2026-09-15 | GHOST | THE EAGLE LANDS AT DAWN | 87% |
JSON line: {"date": "2026-09-15", "agent": "GHOST", "message": "THE EAGLE LANDS AT DAWN", "confidence": 87, "contains_dawn": true}
```
