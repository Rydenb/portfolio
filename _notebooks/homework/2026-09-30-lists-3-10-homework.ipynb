### Popcorn Hack 1: predict, run, fix the crash

The pseudocode uses AP indices, which start at 1. Python starts at 0, so every AP index shifts down by one: AP `items_needed[1]` is Python `items_needed[0]`, and AP `items_needed[LENGTH(items_needed)]` (the last item) is Python `items_needed[len(items_needed) - 1]`.

First, the direct translation of `DISPLAY(items_needed[5])` as `items_needed[4]`. There is no index 4 in a 4-item Python list, so it crashes:

````python
# CODE_RUNNER: Lists Popcorn 1 - predict, run (this version crashes)
items_needed = ["blankets", "socks", "water", "meals"]

print(items_needed[0])                      # prediction: blankets
print(items_needed[len(items_needed) - 1])  # prediction: meals
print(len(items_needed))                    # prediction: 4
print(items_needed[4])                      # prediction: IndexError, there is no 5th item
````

**Actual output:**

```text
blankets
meals
4
IndexError: list index out of range
```

**Fix:** the valid Python indices are 0 to 3, so the last line uses index 3.

````python
# CODE_RUNNER: Lists Popcorn 1 - fixed
items_needed = ["blankets", "socks", "water", "meals"]

print(items_needed[0])                      # prediction: blankets
print(items_needed[len(items_needed) - 1])  # prediction: meals
print(len(items_needed))                    # prediction: 4
print(items_needed[3])                      # prediction: meals (index 3 is the last valid one)
````

**Actual output:**

```text
blankets
meals
4
meals
```

### Popcorn Hack 2: accumulator

**Prediction:** 7 + 6 + 8 + 5 + 7 + 9 = 42, and 42 / 6 = 7.0.

````python
# CODE_RUNNER: Lists Popcorn 2
volunteer_hours = [7, 6, 8, 5, 7, 9]

if len(volunteer_hours) == 0:
    print("No data yet")
else:
    total = 0
    for hours in volunteer_hours:
        total = total + hours
    average = total / len(volunteer_hours)
    print("Total hours:", total)
    print("Average hours:", average)
````

**Actual output:**

```text
Total hours: 42
Average hours: 7.0
```

### Popcorn Hack 3: count then filter

**Prediction:** socks (3) and meals (2) have 2 or more requests.

````python
# CODE_RUNNER: Lists Popcorn 3 - count then filter

items_requested = ["blankets", "socks", "meals", "socks", "water", "socks", "meals"]

frequency = {}
for item in items_requested:
    if item in frequency:
        frequency[item] += 1
    else:
        frequency[item] = 1

# 1. Build a list of items with 2 or more votes
popular = []
for item in frequency:
    if frequency[item] >= 2:
        popular.append(item)

# 2. Print that list
print(popular)
````

**Actual output:**

```text
['socks', 'meals']
```

### Homework: indexing, accumulator, count and filter on one dataset

`total = 0` is set before the accumulator loop, `{}` before the counting loop and `[]` before the filter loop. No `sum()`, no list comprehensions, no hardcoded index for the last donation, and `donations_by_item` is never modified.

````python
# CODE_RUNNER: Lists Homework - indexing, accumulator, count and filter on one dataset

donations_by_item = ["blankets", "socks", "blankets", "water", "socks", "blankets",
                      "meals", "socks", "blankets", "water", "blankets", "socks"]

costs = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]

# LESSON 1 - indexing
# 1. how many donations were made
print(len(donations_by_item))
# 2. the last donation, found with a negative index (no hardcoded 11)
print(donations_by_item[-1])

# LESSON 2 - sum and average
# 3. accumulator loop: total starts at 0 before the loop (no sum())
total = 0
for cost in costs:
    total = total + cost
average = total / len(costs)
# 4. print both as integers
print("Total: " + str(int(total)))
print("Average: " + str(int(average)))

# LESSON 3 - count and filter
# 5. frequency dictionary: start empty, add a key the first time, count up after that
frequency = {}
for item in donations_by_item:
    if item in frequency:
        frequency[item] += 1
    else:
        frequency[item] = 1
print(frequency)

# 6. filter: start with an empty list, append items donated 3 or more times
popular_items = []
for item in frequency:
    if frequency[item] >= 3:
        popular_items.append(item)
print(popular_items)
````

**Actual output:**

```text

socks
Total: 31
Average: 2
{'blankets': 5, 'socks': 4, 'water': 2, 'meals': 1}
['blankets', 'socks']
```
