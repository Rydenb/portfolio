## Homework: Interactive Toggles

Three toggles, each with its own label and purpose (show a message, highlight the panel, show details). JavaScript checks `.checked` on each one, and a live counter shows how many are on. The toggles actually control features, and the code has comments explaining what I learned.

```python
%%html

<!-- UI_RUNNER: Homework - Interactive Toggles -->

<label class="ocs__toggle ocs__toggle--green">
  <input id="notifications" class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Show Message</span>
</label>

<label class="ocs__toggle ocs__toggle--green">
  <input id="highlight" class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Highlight Panel</span>
</label>

<label class="ocs__toggle ocs__toggle--green">
  <input id="details" class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Show Details</span>
</label>

<div id="panel">
  <p id="message" hidden>Notifications are on!</p>
  <p id="detail-text" hidden>Details: each toggle stores its state in a checkbox.</p>
  <p id="count">Toggles on: 0</p>
</div>

<script>
// I learned that .checked lets JavaScript tell whether a toggle is currently on or off.
const messageToggle = document.getElementById("notifications");
const highlightToggle = document.getElementById("highlight");
const detailsToggle = document.getElementById("details");
const toggles = [messageToggle, highlightToggle, detailsToggle];

function updatePage() {
  // Each toggle controls its own feature: show/hide text, or change the background
  document.getElementById("message").hidden = !messageToggle.checked;
  document.getElementById("detail-text").hidden = !detailsToggle.checked;
  document.getElementById("panel").style.background = highlightToggle.checked ? "#FACC15" : "";

  // Count how many toggles are on so the number updates when one is flipped
  let on = 0;
  for (const t of toggles) {
    if (t.checked) {
      on = on + 1;
    }
  }
  document.getElementById("count").textContent = "Toggles on: " + on;
}

// The "change" event fires whenever a toggle is flipped, so the page stays in sync
toggles.forEach(t => t.addEventListener("change", updatePage));
updatePage();
</script>
```
