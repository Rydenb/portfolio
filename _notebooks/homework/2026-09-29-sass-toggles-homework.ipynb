## Homework: Study Dashboard Toggles

Three `ocs__toggle` switches, each with its own label and a real feature:

- **Show study tips** shows and hides a tips section.
- **Highlight the reminder** turns the reminder line into an `ocs__callout`.
- **Open quick links menu** opens and closes a menu of `ocs__btn` links.

The JavaScript reads each toggle's `.checked` value on every `change` event, updates the page, and counts how many toggles are on. The count is in an `<output aria-live="polite">`, so it updates live and screen readers announce it. My insight comments are in the code.

Homework: at least 3 toggles, JavaScript that reacts to each one, and a live count of how many are on.
{% endcapture %}

// Insight: every toggle is really a checkbox, so .checked tells JavaScript
// whether it is on (true) or off (false). The track and circle are only CSS.
const toggles = outputElement.querySelectorAll('.ocs__toggle-input');
const tips = outputElement.querySelector('[data-panel="tips"]');
const reminder = outputElement.querySelector('[data-panel="reminder"]');
const menu = outputElement.querySelector('[data-panel="menu"]');
const count = outputElement.querySelector('[data-count]');

function updatePage() {
  let onCount = 0;
  toggles.forEach(function (toggle) {
    const on = toggle.checked;
    if (on) {
      onCount = onCount + 1;
    }
    // Insight: data-feature connects each toggle to the part of the page it controls,
    // so one function can handle all three instead of writing three copies.
    const feature = toggle.dataset.feature;
    if (feature === 'tips') {
      tips.hidden = !on;          // show/hide a section
    } else if (feature === 'reminder') {
      // classList.toggle adds the class when on is true and removes it when false
      reminder.classList.toggle('ocs__callout', on);
    } else if (feature === 'menu') {
      menu.hidden = !on;          // open/close a menu
    }
  });
  // Insight: aria-live on the output means a screen reader announces the new count.
  count.textContent = onCount;
}

// Question I had: does "change" fire when the label text is clicked instead of the switch?
// Yes: the label wraps the input, so clicking anywhere on it flips the checkbox.
toggles.forEach(function (toggle) {
  toggle.addEventListener('change', updatePage);
});
updatePage();
